```python
import torch
import torch.nn as nn

# 1. Define the Baseline CNN Architecture for Audio Spectrograms
class BaselineAcousticCNN(nn.Module):
    def __init__(self):
        super(BaselineAcousticCNN, self).__init__()

        # Convolutional layer to read the 2D spectrogram array
        self.conv1 = nn.Conv2d(in_channels=1, out_channels=16, kernel_size=3, padding=1)
        self.relu = nn.ReLU()
        self.pool = nn.MaxPool2d(kernel_size=2, stride=2)

        # Flatten and Classify (Real vs. Deepfake)
        # Note: 16 * 64 * 64 assumes a standard 128x128 input size from the preprocessing pipeline
        self.fc1 = nn.Linear(16 * 64 * 64, 2)

    def forward(self, x):
        x = self.conv1(x)
        x = self.relu(x)
        x = self.pool(x)
        x = x.view(x.size(0), -1) # Flattens the arrays for the final layer
        x = self.fc1(x)
        return x

# 2. Initialize the model to prove it works
model = BaselineAcousticCNN()

print("✅ Model architecture successfully built!")
print(model)
```
